# Phase 3 — Trial → Repeat Conversion

**Purpose:** Track what happens after a shopper's trial purchase of アリエールジェル.

| Step | Description |
|------|-------------|
| 3-1 | 6-month cohort funnel by entry size: trial → repeat → lapse |
| 3-2 | Size migration matrix: entry size → next purchase size |
| 3-3 | Pre vs post renewal comparison |
| 3-4 | ASP at trial vs. repeat outcome |

**Definitions (consistent across notebooks 00–05):**
- **Sub-brand Trial** = purchase of ｱﾘｴｰﾙｼﾞｪﾙ with **no purchase of that sub-brand in the prior 12 months** (365 days). NOT first-ever.
- **Repeat** = any subsequent ｱﾘｴｰﾙｼﾞｪﾙ purchase within **180 days** of trial
- **Lapse** = no purchase within **180 days** of trial

### 📏 Canonical Definitions (Cross-Notebook Standard)

| Term | Definition | Window |
|------|-----------|--------|
| **Trial Shopper** | A shopper who purchases the sub-brand/category with **no purchase history of that sub-brand/category in the prior 12 months** (365 days). This is a rolling lookback per purchase event, NOT first-ever. | 365-day lookback |
| **Repeat Shopper** | A trial shopper who makes at least one subsequent purchase of the **same sub-brand** within **180 days** (6 months) after their trial event. | 180-day forward window |
| **Lapsed Shopper** | A trial shopper who makes **no subsequent purchase** of the same sub-brand within **180 days** (6 months) after their trial event. | 180-day forward window |
| **ASP** | `SUM(pos_sales_amt) / SUM(pos_unit_sales_qty)` — weighted average selling price. | Per transaction/aggregation |
| **ASP Band (50 JPY bin)** | `FLOOR(ASP / 50) * 50` — ASP floored to nearest 50 JPY. Label: `ASP (50 JPY bin)`. | — |

**Repeat + Lapse are mutually exclusive and exhaustive** within the trial cohort.

> ⚠️ These definitions are enforced consistently across notebooks 00–05 in this analysis.

## 2. Step 3-1: Build Shopper Cohort — Trial → Repeat → Lapse

```python
# ── Build trial cohort with repeat/lapse classification ───────────────
# DEFINITION UPDATE: Trial = no purchase of same sub-brand in prior 365 days
#   (12-month rolling lookback, NOT first-ever purchase)
# Repeat = subsequent purchase within 180 days of trial
# Lapse  = no purchase within 180 days of trial
#
# Strategy:
# 1. Find each purchase's previous purchase date using LAG()
# 2. A purchase is trial if gap > 365 days (or no prior purchase)
# 3. Find first REPEAT purchase (same sub-brand) within 180 days
# 4. Classify as Repeat or Lapse

cohort_query = f"""
WITH all_purchases AS (
    SELECT
        idpos.shopper_key,
        prod.jp_sub_brand_alter_lang_name AS sub_brand,
        prod.jp_segment_4_name            AS size_code,
        CAST(idpos.sales_period_group_end_date_part AS DATE) AS purchase_date,
        SUM(idpos.pos_sales_amt) / SUM(idpos.pos_unit_sales_qty) AS asp
    FROM cdl_customer_prod.gold_customer_loyalty.loyalty_transact_fct_v1_vw idpos
    LEFT JOIN id_pos_ai_1.prod_dim_ext_vw prod
           ON idpos.prod_key = prod.prod_key
    LEFT JOIN id_pos_ai_1.shopper_dim_generic_vw shopper
           ON idpos.shopper_key = shopper.shopper_key
    WHERE idpos.sales_period_group_end_date_part BETWEEN '{LOOKBACK_START}' AND '{ANALYSIS_END}'
      AND idpos.data_provider_code_part IN ({RETAILER_IN})
      AND prod.jp_category_name = '{CATEGORY}'
      AND prod.jp_sub_category_alter_lang_name = '{SUB_CAT}'
      AND prod.jp_sub_brand_alter_lang_name IN ('{ARIEL_GEL}', '{ATTACK_EX}')
      AND shopper.member_ind = 'Y'
      AND idpos.pos_unit_sales_qty > 0
    GROUP BY 1, 2, 3, 4
),
-- LAG to find previous purchase date per shopper × sub-brand
with_prev AS (
    SELECT *,
        LAG(purchase_date) OVER (
            PARTITION BY shopper_key, sub_brand
            ORDER BY purchase_date
        ) AS prev_purchase_date
    FROM all_purchases
),
-- Trial candidates: purchase in analysis window with >365 day gap (12-month lookback)
trial_candidates AS (
    SELECT *
    FROM with_prev
    WHERE purchase_date BETWEEN '{ANALYSIS_START}' AND '{LATEST_COHORT_END}'
      AND (prev_purchase_date IS NULL
           OR DATEDIFF(purchase_date, prev_purchase_date) > {TRIAL_LOOKBACK_DAYS})
),
-- First trial event per shopper per sub-brand
trial_events AS (
    SELECT
        shopper_key,
        sub_brand,
        MIN(purchase_date) AS trial_date
    FROM trial_candidates
    GROUP BY 1, 2
),
-- Join back to get trial size and ASP
trial_detail AS (
    SELECT
        te.shopper_key,
        te.sub_brand,
        te.trial_date,
        tc.size_code AS trial_size,
        tc.asp       AS trial_asp
    FROM trial_events te
    INNER JOIN trial_candidates tc
           ON te.shopper_key  = tc.shopper_key
          AND te.sub_brand    = tc.sub_brand
          AND te.trial_date   = tc.purchase_date
),
-- Find first repeat purchase (same sub-brand, any size) within 180 days
repeat_events AS (
    SELECT
        td.shopper_key,
        td.sub_brand,
        td.trial_date,
        td.trial_size,
        td.trial_asp,
        MIN(ap.purchase_date)   AS repeat_date,
        MIN(ap.size_code)       AS repeat_size,
        MIN(ap.asp)             AS repeat_asp
    FROM trial_detail td
    LEFT JOIN all_purchases ap
           ON td.shopper_key = ap.shopper_key
          AND td.sub_brand   = ap.sub_brand
          AND ap.purchase_date > td.trial_date
          AND ap.purchase_date <= DATE_ADD(td.trial_date, {REPEAT_LAPSE_WINDOW_DAYS})
    GROUP BY 1, 2, 3, 4, 5
)
SELECT
    re.shopper_key,
    re.sub_brand,
    re.trial_date,
    re.trial_size,
    re.trial_asp,
    re.repeat_date,
    re.repeat_size,
    re.repeat_asp,
    CASE WHEN re.repeat_date IS NOT NULL THEN 'Repeat' ELSE 'Lapse' END AS outcome,
    DATEDIFF(re.repeat_date, re.trial_date) AS days_to_repeat
FROM repeat_events re
ORDER BY re.sub_brand, re.trial_date
"""

print('⏳ Building trial cohort (12-month lookback, this may take a few minutes)...', flush=True)
df_cohort = execute_query(cohort_query)
df_cohort['trial_date'] = pd.to_datetime(df_cohort['trial_date'])
df_cohort['repeat_date'] = pd.to_datetime(df_cohort['repeat_date'])
for col in ['trial_asp', 'repeat_asp', 'days_to_repeat']:
    df_cohort[col] = pd.to_numeric(df_cohort[col])

print(f'\n✅ {len(df_cohort)} shopper cohort records')
print(f'   Ariel: {len(df_cohort[df_cohort["sub_brand"]==ARIEL_GEL]):,}')
print(f'   Attack: {len(df_cohort[df_cohort["sub_brand"]==ATTACK_EX]):,}')
print(f'   ℹ️  Trial = no sub-brand purchase in prior {TRIAL_LOOKBACK_DAYS} days')
print(f'   ℹ️  Repeat/Lapse window = {REPEAT_LAPSE_WINDOW_DAYS} days')
```

```python
# ── Cohort funnel summary by entry size ───────────────────────────────
df_cohort['trial_month'] = df_cohort['trial_date'].dt.to_period('M').astype(str)

funnel = df_cohort.groupby(['sub_brand', 'trial_size', 'outcome']).agg(
    shoppers=('shopper_key', 'nunique')
).reset_index()

funnel_pivot = funnel.pivot_table(
    index=['sub_brand', 'trial_size'],
    columns='outcome',
    values='shoppers',
    fill_value=0
).reset_index()

if 'Repeat' in funnel_pivot.columns and 'Lapse' in funnel_pivot.columns:
    funnel_pivot['total'] = funnel_pivot['Repeat'] + funnel_pivot['Lapse']
    funnel_pivot['repeat_rate_%'] = (funnel_pivot['Repeat'] / funnel_pivot['total'] * 100).round(1)
    funnel_pivot['lapse_rate_%'] = (funnel_pivot['Lapse'] / funnel_pivot['total'] * 100).round(1)

print('=' * 80)
print('6-Month Cohort Funnel: Trial → Repeat / Lapse by Entry Size')
print(f'Trial cohorts: {ANALYSIS_START} → {LATEST_COHORT_END}')
print('=' * 80)

print('\n▶ アリエールジェル')
print(funnel_pivot[funnel_pivot['sub_brand'] == ARIEL_GEL].to_string(index=False))

print('\n▶ アタック抗菌EX')
print(funnel_pivot[funnel_pivot['sub_brand'] == ATTACK_EX].to_string(index=False))
```

```python
# ── Visualize repeat rate by size ─────────────────────────────────────
if 'repeat_rate_%' in funnel_pivot.columns:
    # Apply SIZE_ORDER and exclude EXCLUDED_SIZES
    valid_sizes = order_and_filter_sizes(funnel_pivot['trial_size'].unique())
    plot_df = funnel_pivot[funnel_pivot['trial_size'].isin(valid_sizes)].copy()
    plot_df['trial_size'] = pd.Categorical(plot_df['trial_size'], categories=valid_sizes, ordered=True)
    plot_df = plot_df.sort_values('trial_size')

    fig = px.bar(plot_df, x='trial_size', y='repeat_rate_%', color='sub_brand',
                 barmode='group', color_discrete_map={ARIEL_GEL: '#1E90FF', ATTACK_EX: '#FF6347'},
                 text='repeat_rate_%',
                 labels={'repeat_rate_%': 'Repeat Rate (%)', 'trial_size': 'Trial Entry Size'},
                 title='6-Month Repeat Rate by Trial Entry Size — Higher = Better Retention',
                 category_orders={'trial_size': valid_sizes})
    fig.update_traces(texttemplate='%{text:.1f}%', textposition='outside')
    fig.update_layout(template='plotly_white', height=500)
    fig.show()
```

## 3. Step 3-2: Size Migration Matrix

```python
# ── Size migration: trial size → first repeat size (Ariel) ────────────
repeat_shoppers = df_cohort[
    (df_cohort['outcome'] == 'Repeat') &
    (df_cohort['sub_brand'] == ARIEL_GEL) &
    (~df_cohort['trial_size'].isin(EXCLUDED_SIZES)) &
    (~df_cohort['repeat_size'].isin(EXCLUDED_SIZES))
].copy()

if len(repeat_shoppers) > 0:
    migration = repeat_shoppers.groupby(['trial_size', 'repeat_size']).agg(
        shoppers=('shopper_key', 'nunique')
    ).reset_index()

    # Pivot
    migration_matrix = migration.pivot_table(
        index='trial_size', columns='repeat_size', values='shoppers', fill_value=0
    )
    # Reorder axes by SIZE_ORDER
    ordered_sizes = order_and_filter_sizes(
        list(migration_matrix.index) + list(migration_matrix.columns)
    )
    idx_ordered  = [s for s in ordered_sizes if s in migration_matrix.index]
    col_ordered  = [s for s in ordered_sizes if s in migration_matrix.columns]
    migration_matrix = migration_matrix.reindex(index=idx_ordered, columns=col_ordered, fill_value=0)

    # Add percentages
    migration_pct = migration_matrix.div(migration_matrix.sum(axis=1), axis=0) * 100

    print('=' * 80)
    print('Size Migration Matrix — アリエールジェル (Trial → Repeat)')
    print('Rows = Trial entry size | Columns = First repeat size | Values = % of trial shoppers')
    print('=' * 80)
    print(migration_pct.round(1).to_string())

    print('\nAbsolute counts:')
    print(migration_matrix.to_string())

    # ── Heatmap ────────────────────────────────────────────────────────
    fig = px.imshow(migration_pct.values,
                    x=migration_pct.columns.tolist(),
                    y=migration_pct.index.tolist(),
                    text_auto='.1f', color_continuous_scale='Blues',
                    labels=dict(x='Repeat Size', y='Trial Size', color='%'),
                    title='[アリエールジェル] Size Migration Heatmap: Where Do Trial Shoppers Repeat? (%)')
    fig.update_layout(height=500, template='plotly_white')
    fig.show()
else:
    print('⚠️ No Ariel repeat shoppers found')
```

```python
# ── (NEW) Size migration: アタック抗菌EX ─────────────────────────────
attack_repeat_shoppers = df_cohort[
    (df_cohort['outcome'] == 'Repeat') &
    (df_cohort['sub_brand'] == ATTACK_EX) &
    (~df_cohort['trial_size'].isin(EXCLUDED_SIZES)) &
    (~df_cohort['repeat_size'].isin(EXCLUDED_SIZES))
].copy()

if len(attack_repeat_shoppers) > 0:
    atk_migration = attack_repeat_shoppers.groupby(['trial_size', 'repeat_size']).agg(
        shoppers=('shopper_key', 'nunique')
    ).reset_index()

    atk_matrix = atk_migration.pivot_table(
        index='trial_size', columns='repeat_size', values='shoppers', fill_value=0
    )
    atk_ordered = order_and_filter_sizes(
        list(atk_matrix.index) + list(atk_matrix.columns)
    )
    atk_idx = [s for s in atk_ordered if s in atk_matrix.index]
    atk_col = [s for s in atk_ordered if s in atk_matrix.columns]
    atk_matrix = atk_matrix.reindex(index=atk_idx, columns=atk_col, fill_value=0)

    atk_pct = atk_matrix.div(atk_matrix.sum(axis=1), axis=0) * 100

    print('=' * 80)
    print('Size Migration Matrix — アタック抗菌EX (Trial → Repeat)')
    print('Rows = Trial entry size | Columns = First repeat size | Values = % of trial shoppers')
    print('=' * 80)
    print(atk_pct.round(1).to_string())

    fig_atk = px.imshow(atk_pct.values,
                        x=atk_pct.columns.tolist(),
                        y=atk_pct.index.tolist(),
                        text_auto='.1f', color_continuous_scale='Oranges',
                        labels=dict(x='Repeat Size', y='Trial Size', color='%'),
                        title='[アタック抗菌EX] Size Migration Heatmap: Where Do Trial Shoppers Repeat? (%)')
    fig_atk.update_layout(height=500, template='plotly_white')
    fig_atk.show()
else:
    print('⚠️ No Attack repeat shoppers found')
```

```python

# ── (NEW) Migration direction comparison: Ariel Gel vs Attack 抗菌EX ─────
# Classify each trial→repeat transition as:  Same Size / Size Up / Size Down
# Shows which brand retains shoppers in the same SKU vs prompts trade-up/down

def classify_migration(trial_size, repeat_size, size_order):
    """Return migration direction based on SIZE_ORDER index."""
    if trial_size not in size_order or repeat_size not in size_order:
        return 'Other'
    ti, ri = size_order.index(trial_size), size_order.index(repeat_size)
    if ti == ri:
        return '① Same Size'
    elif ri > ti:
        return '③ Size Up (Larger)'
    else:
        return '② Size Down (Smaller)'

MIGRATE_ORDER  = ['① Same Size', '② Size Down (Smaller)', '③ Size Up (Larger)']
MIGRATE_COLORS = {
    '① Same Size':         '#4ECDC4',
    '② Size Down (Smaller)': '#E74C3C',
    '③ Size Up (Larger)':  '#2ECC71',
}

dfs_mig = []
for brand, df_brand in [(ARIEL_GEL, repeat_shoppers), (ATTACK_EX, attack_repeat_shoppers)]:
    if len(df_brand) == 0:
        continue
    tmp = df_brand[
        (~df_brand['trial_size'].isin(EXCLUDED_SIZES)) &
        (~df_brand['repeat_size'].isin(EXCLUDED_SIZES))
    ].copy()
    tmp['migration_type'] = tmp.apply(
        lambda row: classify_migration(row['trial_size'], row['repeat_size'], SIZE_ORDER), axis=1
    )
    tmp_agg = tmp.groupby(['trial_size', 'migration_type']).agg(
        shoppers=('shopper_key', 'nunique')
    ).reset_index()
    tmp_agg['brand'] = brand
    dfs_mig.append(tmp_agg)

if dfs_mig:
    mig_compare = pd.concat(dfs_mig, ignore_index=True)
    mig_compare = mig_compare[mig_compare['migration_type'] != 'Other']

    # % within each brand × trial_size bucket
    totals = mig_compare.groupby(['brand', 'trial_size'])['shoppers'].transform('sum')
    mig_compare['pct'] = (mig_compare['shoppers'] / totals * 100).round(1)

    # Apply SIZE_ORDER ordering
    valid_sizes = order_and_filter_sizes(mig_compare['trial_size'].unique())
    mig_compare['trial_size'] = pd.Categorical(mig_compare['trial_size'], categories=valid_sizes, ordered=True)
    mig_compare = mig_compare[mig_compare['trial_size'].notna()].sort_values('trial_size')

    brands_list = [ARIEL_GEL, ATTACK_EX]
    brands_available = [b for b in brands_list if b in mig_compare['brand'].values]

    fig_mig = make_subplots(
        rows=len(brands_available), cols=1,
        subplot_titles=brands_available,
        vertical_spacing=0.18
    )

    for row_idx, brand in enumerate(brands_available, 1):
        bd = mig_compare[mig_compare['brand'] == brand]
        for mtype in MIGRATE_ORDER:
            sub = bd[bd['migration_type'] == mtype]
            if len(sub) == 0:
                continue
            fig_mig.add_trace(go.Bar(
                x=sub['trial_size'].astype(str),
                y=sub['pct'],
                name=mtype,
                marker_color=MIGRATE_COLORS[mtype],
                text=sub['pct'].apply(lambda v: f'{v:.1f}%'),
                textposition='inside',
                insidetextanchor='middle',
                legendgroup=mtype,
                showlegend=(row_idx == 1)
            ), row=row_idx, col=1)

    fig_mig.update_layout(
        height=420 * len(brands_available),
        title_text=(
            'Size Migration Direction: Ariel Gel vs Attack 抗菌EX<br>'
            '<sup>% of repeat shoppers — Same size vs Size Down (smaller) vs Size Up (larger refill)</sup>'
        ),
        barmode='stack',
        template='plotly_white',
        legend=dict(orientation='h', y=-0.05)
    )
    fig_mig.update_yaxes(title_text='% of Repeat Shoppers', range=[0, 105])
    fig_mig.update_xaxes(title_text='Trial Entry Size')
    fig_mig.show()

    # ── Summary table ──────────────────────────────────────────────────
    print('\n▶ Migration Direction Summary:')
    summary_tbl = mig_compare.pivot_table(
        index=['brand', 'trial_size'],
        columns='migration_type',
        values='pct',
        fill_value=0
    ).round(1)
    print(summary_tbl.to_string())
else:
    print('⚠️ No migration data available')
```

## 4. Step 3-3: Pre vs Post Renewal Repeat Rate Comparison

```python
# ── Split cohorts into pre-renewal and post-renewal ───────────────────
renewal_date = pd.Timestamp(RENEWAL_MONTH)
df_cohort['renewal_period'] = df_cohort['trial_date'].apply(
    lambda x: 'Pre-Renewal' if x < renewal_date else 'Post-Renewal'
)

period_funnel = df_cohort[df_cohort['sub_brand'] == ARIEL_GEL].groupby(
    ['renewal_period', 'trial_size', 'outcome']
).agg(shoppers=('shopper_key', 'nunique')).reset_index()

period_pivot = period_funnel.pivot_table(
    index=['renewal_period', 'trial_size'],
    columns='outcome',
    values='shoppers',
    fill_value=0
).reset_index()

if 'Repeat' in period_pivot.columns and 'Lapse' in period_pivot.columns:
    period_pivot['total'] = period_pivot['Repeat'] + period_pivot['Lapse']
    period_pivot['repeat_rate_%'] = (period_pivot['Repeat'] / period_pivot['total'] * 100).round(1)

    print('=' * 80)
    print('Pre vs Post Renewal: Repeat Rate by Trial Size — アリエールジェル')
    print('=' * 80)
    print(period_pivot.to_string(index=False))

    # ── Grouped bar chart ──────────────────────────────────────────────
    fig = px.bar(period_pivot, x='trial_size', y='repeat_rate_%', color='renewal_period',
                 barmode='group', text='repeat_rate_%',
                 color_discrete_map={'Pre-Renewal': '#6495ED', 'Post-Renewal': '#FF6347'},
                 title='Did the Renewal Improve Repeat Rates? (by Trial Entry Size)')
    fig.update_traces(texttemplate='%{text:.1f}%', textposition='outside')
    fig.update_layout(template='plotly_white', height=500,
                      yaxis_title='Repeat Rate (%)')
    fig.show()
```

## 5. Step 3-4: Trial ASP vs Repeat Outcome

```python

# ── ASP 50JPY bin vs Trial Shoppers + Frequency — Ariel (by size) ────
# Frequency (purchase occasions) shown as a line on the secondary Y-axis.
# Also produces separate Repeat-shopper and Lapse-shopper versions.

# ── Base data including outcome ────────────────────────────────────────
ariel_trials = df_cohort[
    (df_cohort['sub_brand'] == ARIEL_GEL) &
    (~df_cohort['trial_size'].isin(EXCLUDED_SIZES))
].copy()
ariel_trials['asp_band'] = (ariel_trials['trial_asp'] // 50 * 50).astype('Int64')

# Aggregate: all trial (regardless of outcome) + split by outcome
asp_bin_agg = ariel_trials.groupby(['trial_size', 'asp_band']).agg(
    trial_shoppers=('shopper_key', 'nunique'),
    obs_count=('shopper_key', 'count')
).reset_index()

asp_bin_agg_split = ariel_trials.groupby(['trial_size', 'asp_band', 'outcome']).agg(
    shoppers=('shopper_key', 'nunique'),
    obs_count=('shopper_key', 'count')
).reset_index()

# Frequency filter: 25th-pct threshold (min=5) to skip low-data bands
# ⚠️ INTENTIONAL EXCEPTION: Uses quantile-based threshold (min=5) because
#    trial cohort counts are smaller than NB04's all-purchaser base (MIN_FREQ=30).
#    NB02 uses MIN_BIN_OBS=2 at weekly granularity for different reasons.
def filter_freq(grp, pct=25):
    threshold = grp['obs_count'].quantile(pct / 100)
    return grp[grp['obs_count'] >= max(threshold, 5)]

asp_bin_filtered = (
    asp_bin_agg.groupby('trial_size', group_keys=False)
    .apply(filter_freq)
    .reset_index(drop=True)
)

asp_bin_filtered_split = (
    asp_bin_agg_split.groupby(['trial_size', 'outcome'], group_keys=False)
    .apply(filter_freq)
    .reset_index(drop=True)
)

sizes_asp = order_and_filter_sizes(asp_bin_filtered['trial_size'].unique())

def make_asp_bin_chart(data_df, y_col, obs_col, sizes, title, y_label):
    """Bar chart of shoppers by ASP band; frequency shown as a line on secondary Y-axis."""
    if len(sizes) == 0:
        print(f'⚠️ No data for: {title}')
        return
    n_s = len(sizes)
    n_c = min(2, n_s)
    n_r = (n_s + n_c - 1) // n_c

    specs = [[{"secondary_y": True} for _ in range(n_c)] for _ in range(n_r)]

    fig = make_subplots(
        rows=n_r, cols=n_c,
        specs=specs,
        subplot_titles=[f'Ariel {s}' for s in sizes],
        vertical_spacing=0.16, horizontal_spacing=0.14
    )

    for idx, size in enumerate(sizes):
        r = idx // n_c + 1
        c = idx % n_c + 1
        subset = data_df[data_df['trial_size'] == size].sort_values('asp_band')
        if len(subset) == 0:
            continue

        x_labels = subset['asp_band'].astype(str) + '~'

        # ── Bar: shopper count (primary Y) ────────────────────────────
        fig.add_trace(go.Bar(
            x=x_labels,
            y=subset[y_col],
            name=y_label,
            marker_color='#4C9BE8',
            opacity=0.8,
            showlegend=(idx == 0),
            legendgroup='shoppers'
        ), row=r, col=c, secondary_y=False)

        # ── Line: purchase occasions (secondary Y) ────────────────────
        fig.add_trace(go.Scatter(
            x=x_labels,
            y=subset[obs_col],
            name='Purchase Occasions',
            mode='lines+markers',
            line=dict(color='#FF6347', width=2),
            marker=dict(size=6, symbol='circle'),
            showlegend=(idx == 0),
            legendgroup='occasions'
        ), row=r, col=c, secondary_y=True)

    fig.update_layout(
        height=420 * n_r,
        title_text=title,
        template='plotly_white',
        legend=dict(orientation='h', y=-0.04, x=0.5, xanchor='center')
    )
    fig.update_xaxes(title_text='ASP (50 JPY bin)')
    for row in range(1, n_r + 1):
        for col in range(1, n_c + 1):
            fig.update_yaxes(title_text=y_label,              secondary_y=False, row=row, col=col)
            fig.update_yaxes(title_text='Purchase Occasions', secondary_y=True,  row=row, col=col,
                             showgrid=False)
    fig.show()


# ── Chart 1: All trial shoppers ────────────────────────────────────────
make_asp_bin_chart(
    data_df=asp_bin_filtered,
    y_col='trial_shoppers',
    obs_col='obs_count',
    sizes=sizes_asp,
    title=(
        'アリエールジェル: Trial Entry ASP Band (50JPY) vs Trial Shoppers — ALL<br>'
        '<sup>Blue bars = trial shoppers (left axis) | Orange line = purchase occasions / frequency (right axis)</sup>'
    ),
    y_label='Trial Shoppers'
)

# ── Chart 2: Repeat shoppers only ─────────────────────────────────────
repeat_asp = asp_bin_filtered_split[asp_bin_filtered_split['outcome'] == 'Repeat']
sizes_repeat = order_and_filter_sizes(repeat_asp['trial_size'].unique())
make_asp_bin_chart(
    data_df=repeat_asp,
    y_col='shoppers',
    obs_col='obs_count',
    sizes=sizes_repeat,
    title=(
        'アリエールジェル: Trial ASP Band vs Shoppers — REPEAT only (6-month repeat)<br>'
        '<sup>Blue bars = repeat shoppers (left axis) | Orange line = purchase occasions / frequency (right axis)</sup>'
    ),
    y_label='Repeat Shoppers'
)

# ── Chart 3: Lapse shoppers only ───────────────────────────────────────
lapse_asp = asp_bin_filtered_split[asp_bin_filtered_split['outcome'] == 'Lapse']
sizes_lapse = order_and_filter_sizes(lapse_asp['trial_size'].unique())
make_asp_bin_chart(
    data_df=lapse_asp,
    y_col='shoppers',
    obs_col='obs_count',
    sizes=sizes_lapse,
    title=(
        'アリエールジェル: Trial ASP Band vs Shoppers — LAPSE only (no repeat within 6 months)<br>'
        '<sup>Blue bars = lapse shoppers (left axis) | Orange line = purchase occasions / frequency (right axis)</sup>'
    ),
    y_label='Lapse Shoppers'
)
```

```python

# ── Step 3-4b: Trial / Repeat / Lapse RATE by ASP Band ────────────────
# Denominator = all Ariel Gel shoppers who purchased at that ASP band
# (same analysis window: ANALYSIS_START → ANALYSIS_END)
#
# Trial rate   = trial shoppers ÷ all shoppers at that ASP band
# Repeat rate  = repeat shoppers ÷ trial shoppers at that ASP band
# Lapse rate   = lapse shoppers  ÷ trial shoppers at that ASP band
# Note: Repeat rate + Lapse rate ≈ 100% (within-trial); Trial rate is independent

# ── 1. Query: all Ariel Gel shoppers per (size, ASP 50JPY band) ───────
# CTE first computes per-transaction ASP (avoiding aggregate-in-GROUP-BY error),
# then bins it and counts distinct shoppers.
universe_query = f"""
WITH txn_asp AS (
    SELECT
        idpos.shopper_key,
        prod.jp_segment_4_name                                          AS trial_size,
        CAST(
            FLOOR(
                (SUM(idpos.pos_sales_amt) / SUM(idpos.pos_unit_sales_qty)) / 50
            ) * 50
        AS BIGINT)                                                       AS asp_band
    FROM cdl_customer_prod.gold_customer_loyalty.loyalty_transact_fct_v1_vw idpos
    LEFT JOIN id_pos_ai_1.prod_dim_ext_vw prod
           ON idpos.prod_key = prod.prod_key
    LEFT JOIN id_pos_ai_1.shopper_dim_generic_vw shopper
           ON idpos.shopper_key = shopper.shopper_key
    WHERE idpos.sales_period_group_end_date_part BETWEEN '{ANALYSIS_START}' AND '{ANALYSIS_END}'
      AND idpos.data_provider_code_part IN ({RETAILER_IN})
      AND prod.jp_category_name = '{CATEGORY}'
      AND prod.jp_sub_category_alter_lang_name = '{SUB_CAT}'
      AND prod.jp_sub_brand_alter_lang_name = '{ARIEL_GEL}'
      AND prod.jp_segment_4_name NOT IN ({', '.join(f"'{s}'" for s in EXCLUDED_SIZES)})
      AND shopper.member_ind = 'Y'
      AND idpos.pos_unit_sales_qty > 0
    GROUP BY
        idpos.shopper_key,
        prod.jp_segment_4_name,
        idpos.sales_period_group_end_date_part
)
SELECT
    trial_size,
    asp_band,
    COUNT(DISTINCT shopper_key) AS all_shoppers
FROM txn_asp
GROUP BY trial_size, asp_band
"""

print('⏳ Fetching Ariel Gel shopper universe by ASP band...', flush=True)
universe_df = execute_query(universe_query)
universe_df['asp_band'] = pd.to_numeric(universe_df['asp_band'])
universe_df['all_shoppers'] = pd.to_numeric(universe_df['all_shoppers'])
print(f'✅ {len(universe_df)} (size × ASP band) rows fetched')

# ── 2. Build rate table ────────────────────────────────────────────────
# trial counts per (size, band) from asp_bin_filtered (already freq-filtered)
trial_counts = asp_bin_filtered[['trial_size', 'asp_band', 'trial_shoppers', 'obs_count']].copy()

# repeat / lapse counts per (size, band)
repeat_counts = (
    asp_bin_filtered_split[asp_bin_filtered_split['outcome'] == 'Repeat']
    [['trial_size', 'asp_band', 'shoppers']].rename(columns={'shoppers': 'repeat_shoppers'})
)
lapse_counts = (
    asp_bin_filtered_split[asp_bin_filtered_split['outcome'] == 'Lapse']
    [['trial_size', 'asp_band', 'shoppers']].rename(columns={'shoppers': 'lapse_shoppers'})
)

# merge everything
rate_df = (
    trial_counts
    .merge(universe_df, on=['trial_size', 'asp_band'], how='left')
    .merge(repeat_counts, on=['trial_size', 'asp_band'], how='left')
    .merge(lapse_counts,  on=['trial_size', 'asp_band'], how='left')
)
rate_df[['repeat_shoppers', 'lapse_shoppers']] = (
    rate_df[['repeat_shoppers', 'lapse_shoppers']].fillna(0)
)

# compute rates
rate_df['trial_rate_%']  = (rate_df['trial_shoppers']  / rate_df['all_shoppers']  * 100).round(1)
rate_df['repeat_rate_%'] = (rate_df['repeat_shoppers'] / rate_df['trial_shoppers'] * 100).round(1)
rate_df['lapse_rate_%']  = (rate_df['lapse_shoppers']  / rate_df['trial_shoppers'] * 100).round(1)

# cap at 100% (rounding artefacts)
for col in ['trial_rate_%', 'repeat_rate_%', 'lapse_rate_%']:
    rate_df[col] = rate_df[col].clip(upper=100)

print('\n▶ Rate table sample:')
print(rate_df[['trial_size', 'asp_band', 'all_shoppers', 'trial_shoppers',
               'trial_rate_%', 'repeat_rate_%', 'lapse_rate_%']].head(12).to_string(index=False))

# ── 3. Visualize: bars (all shoppers) + 3 rate lines per size ─────────
sizes_rate = order_and_filter_sizes(rate_df['trial_size'].unique())

if sizes_rate:
    n_s = len(sizes_rate)
    n_c = min(2, n_s)
    n_r = (n_s + n_c - 1) // n_c

    specs = [[{"secondary_y": True} for _ in range(n_c)] for _ in range(n_r)]

    fig_rate = make_subplots(
        rows=n_r, cols=n_c,
        specs=specs,
        subplot_titles=[f'Ariel {s}' for s in sizes_rate],
        vertical_spacing=0.16, horizontal_spacing=0.14
    )

    RATE_COLORS = {
        'trial_rate_%':  '#1E90FF',   # blue  — trial
        'repeat_rate_%': '#2ECC71',   # green — repeat
        'lapse_rate_%':  '#E74C3C',   # red   — lapse
    }
    RATE_LABELS = {
        'trial_rate_%':  'Trial Rate (% of all buyers)',
        'repeat_rate_%': 'Repeat Rate (% of trialists)',
        'lapse_rate_%':  'Lapse Rate (% of trialists)',
    }

    for idx, size in enumerate(sizes_rate):
        r = idx // n_c + 1
        c = idx % n_c + 1
        subset = rate_df[rate_df['trial_size'] == size].sort_values('asp_band')
        if len(subset) == 0:
            continue

        x_labels = subset['asp_band'].astype(str) + '~'

        # Bar: total shopper volume (primary Y = denominator context)
        fig_rate.add_trace(go.Bar(
            x=x_labels,
            y=subset['all_shoppers'],
            name='All Shoppers (volume)',
            marker_color='#D0D0D0',
            opacity=0.6,
            showlegend=(idx == 0),
            legendgroup='all_vol'
        ), row=r, col=c, secondary_y=False)

        # Lines: 3 rates (secondary Y, 0–100%)
        for rate_col, color in RATE_COLORS.items():
            fig_rate.add_trace(go.Scatter(
                x=x_labels,
                y=subset[rate_col],
                name=RATE_LABELS[rate_col],
                mode='lines+markers',
                line=dict(color=color, width=2),
                marker=dict(size=6),
                showlegend=(idx == 0),
                legendgroup=rate_col
            ), row=r, col=c, secondary_y=True)

    fig_rate.update_layout(
        height=440 * n_r,
        title_text=(
            'アリエールジェル: ASP Band — Trial / Repeat / Lapse Rates (by Entry Size)<br>'
            '<sup>'
            'Grey bars = total buyers at that ASP band (left axis) | '
            'Blue = Trial rate (% of all buyers) | '
            'Green = Repeat rate (% of trialists) | '
            'Red = Lapse rate (% of trialists)'
            '</sup>'
        ),
        template='plotly_white',
        legend=dict(orientation='h', y=-0.04, x=0.5, xanchor='center')
    )
    fig_rate.update_xaxes(title_text='ASP Band (JPY)')
    for _r in range(1, n_r + 1):
        for _c in range(1, n_c + 1):
            fig_rate.update_yaxes(title_text='All Shoppers',  secondary_y=False, row=_r, col=_c)
            fig_rate.update_yaxes(title_text='Rate (%)',      secondary_y=True,  row=_r, col=_c,
                                  range=[0, 105], showgrid=False)
    fig_rate.show()
else:
    print('⚠️ No rate data to plot')
```
